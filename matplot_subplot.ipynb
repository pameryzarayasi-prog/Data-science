```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.figure import Figure
```

```python
fig =plt.figure()
```

```python
x=np.linspace(0,5,11)
y1=x**2
y2=x**3
```

```python
fig =plt.figure()
ax=fig.add_subplot()
ax.plot(x,y1, "ro-")
ax.set_title("Funcion cuadratica")
plt.show()
```

```python
fig2 =plt.figure()
ax=fig2.add_subplot()
ax.plot(x,y2, "go-")
ax.set_title("Funcion cúbica")
plt.show()
```

```python
dashboard = plt.figure()
axes1 = dashboard.add_axes([0.1, 0.1, 0.8, 0.8])

# Trazar ambas funciones en axes1
axes1.plot(x, y1, "b-", label="Cuadrática")
axes1.plot(x, y2, "go-", label="Funcion cúbica")

axes1.set_title("Dashboard Unificado")
axes1.legend()

plt.show()
```

```python

# 2. Crear la figura principal
dashboard = plt.figure(figsize=(10, 4)) # figsize da espacio horizontal

# 3. Agregar los ejes uno al lado del otro
axes1 = dashboard.add_subplot(121)  # Lado izquierdo
axes2 = dashboard.add_subplot(122)  # Lado derecho

# 4. Graficar en cada eje
axes1.plot(x, y1, "b-", label="Cuadrática")
axes2.plot(x, y2, "go-", label="Cúbica")

# 5. Títulos y ajustes
axes1.set_title("Dashboard Principal")
axes2.set_title("Funcion cúbica")

plt.tight_layout() # Evita que se solapen los títulos y ejes
plt.show()
```

```python
x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3
```

```python
def grafico_cuadratico(ax, x, y):
    ax.plot(x, y, 'ro--')
    ax.set_title('Función cuadrática')
    ax.grid(True, linestyle='--', axis='both', gapcolor='green', linewidth=2)
    ax.set_xlabel('Input')
    ax.set_ylabel('Output')


def grafico_cubico(ax, x, y):
    ax.plot(x, y, 'bo-')
    ax.set_title('Función cúbica')
    ax.grid(True)

def grafico_pastel(ax, pct, labels):
    ax.pie(pct, labels=labels, autopct='%1.1f%%')
    ax.set_title('Gráfico Pastel')


dashboard, axes = plt.subplots(1, 2, figsize=(12, 5))

grafico_cuadratico(axes[0], x, y1)
grafico_cubico(axes[1], x, y2)

dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```

```python
frutas = ['Manzanas', 'Plátanos', 'Peras']
porcentajes = [30, 50, 20]


dashboard, axes = plt.subplots(1, 3, figsize=(12, 5))

grafico_cuadratico(axes[0], x, y1)
grafico_pastel(axes[1], porcentajes, frutas)
grafico_cubico(axes[2], x, y2)


dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```

```python
import numpy as np

plt.style.use('classic')

x = np.linspace(1, 20, 100)

dashboard, axes = plt.subplots(2, 4, figsize=(16, 8))

axes[0, 0].plot(x, np.cos(x))

axes[1, 3].plot(x, np.sin(x), 'r--', label='f(x) = sin(x)')
axes[1, 3].set_title("Onda sinusoidal")
axes[1, 3].set_xlabel('Muestras (ms)')
axes[1, 3].set_ylabel('Amplitud')
axes[1, 3].legend(loc='upper right')
axes[1, 3].grid(True, color='blue', linestyle='--', alpha=0.4)


axes[0, 2].plot(x, np.tanh(x), 'r--', label='f(x) = tanh(x)')
axes[0, 2].set_title("Onda Tangente Hiperbólica")
axes[0, 2].set_xlabel('Muestras (ms)')
axes[0, 2].set_ylabel('Amplitud')
axes[0, 2].legend(loc='upper right')
axes[0, 2].grid(True, color='orange', linestyle='--', alpha=0.4)

dashboard.tight_layout(h_pad=2.5)
plt.show()
```
