```python
#1 Analizar las ventas de una tienda
#Crea una función llamada analizar_ventas(ventas) que reciba una lista de montos de venta y construya un resumen estadístico.
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
#analizar_ventas(ventas)={}

def analizar_ventas(ventas: list[int | float]) -> dict[str, int | float]:
    if not ventas:
        return "tiene un error"
    
    for venta in ventas:
        if venta<0:
            return f"Hay ventas negativas: {venta}"
    
    total=sum(ventas)
    promedio=sum(ventas)/len(ventas)
    venta_mas_alta= max(ventas)
    venta_mas_baja= min(ventas)

    venta_superior_prom= 0
    venta_inferior_100=0

    for venta in ventas:
        if venta>promedio:
            venta_superior_prom+=1
        if venta<100:
            venta_inferior_100+=1

    return{
        "total": total,
        "promedio": promedio,
        "venta_mas_alta": venta_mas_alta,
        "venta_mas_baja": venta_mas_baja,
        "venta_superior_prom": venta_superior_prom,
        "venta_inferior_100": venta_inferior_100
    }
analizar_ventas(ventas)
```

```python
#2 Validar una contraseña y explicar los errores
#Enunciado: Crea una función validar_password(password) que determine si una contraseña cumple todas las reglas de seguridad.

#def analizar_ventas(ventas: list[int | float]) -> dict[str, int | float]:
password=input("ingrese contraseña")

def validar_password(password: str):
    simbolo = ["@", "#", "$", "o", "%" ]
    errores=[]

    if len(password)<8:
        errores.append("Debe tener al menos 8 caracteres")

    contiene_minus = False
    contiene_mayus = False
    contien_numero = False

    for caracter in password:
        if caracter.islower():
            contiene_minus=True
            break
    for caracter in password:
        if caracter.isupper():
            contiene_minus=True
            break
    for caracter in password:
        if caracter.isnumeric():
            contien_numero=True
            break   
    if not contiene_minus:
        errores.append(
            "No contiene minuscula"
        )
    if not contiene_mayus:
        errores.append(
            "No contiene MAYUSCULA"
        )
    if not contien_numero:
        errores.append(
            "No contiene numero"
        )
    if not simbolo:
        errores.append(
            "No contiene simbolo"
        )
    if len(password)!= len(password.strip()):
        errores.append(
            "Existe espacios"
        )
    valido=False if errores else True
    return(valido,errores)
validar_password(password)
```

```python
#3 Registrar estudiantes con validaciones
#Enunciado: Crea una función registrar_estudiante(estudiantes, nombre, notas) que agregue la informaciónde un estudiante a un diccionario.
def  registrar_estudiante(
    estudiantes:dict[str,int|float|str],
    nombre:str,
    notas:list[int|float]
) -> str:
    
    if not nombre:
        return "Nombre invalido"
    
    if nombre in estudiantes:
        return "El nombre existe"
    
    for nota in notas:
        if isinstance(nota, str):
            return "hay notas no numericas"
        else:
            if nota<0 or nota>20:
                return "Nota invalida"
    
    promedio = round(sum(notas)/len(notas),2)
    
    if promedio>=11:
        estado = "Aprobado"
    else:
        estado= "Desaprobado"
    
    estudiantes[nombre]={
        "estado": estado,
        "promedio": promedio,
        "notas": notas
    }
    return "registro exitoso"
```

```python
#4 Agrupar palabras por longitud
#Crea una función agrupar_por_longitud(palabras) que organice una lista de palabras dentro de un diccionario

palabras = [ 
'python', 'Java', 'código', 'web', 
'java', 'datos', 'API', 'lista' 
] 

def agrupar_por_longitud(palabras:list[str])->dict[int,list[str]]:
    palabras_limpias=set([
            palabra.lower()
            for palabra in palabras
            if isinstance(palabra,str)
    ])
    grupos={}
    for palabras_limpias in palabras_limpias:
        longitud=len(palabras_limpias)
        grupos.setdefault(longitud,[]).append(palabras_limpias) #setdefault crea una nueva lista en caso no exista
    
    print(grupos)
    
    for llave in grupos:
        grupos[llave].sort()
    
    print(grupos)

    return(grupos)

print(agrupar_por_longitud(palabras))
agrupar_por_longitud(palabras)
```

```python
#5
inventario = { 
'Laptop': 5, 
'Mouse': 12, 
'Teclado': 8 
} 
#actualizar_inventario(inventario, 'Mouse', 4, 'salida'):

def actualizar_inventario(
        inventario:dict[str,int], 
        producto:str, 
        cantidad:int, 
        operacion:str
        )->dict[str,str|int]:
        operaciones = ( "entrada", "salida")
        if operacion.lower() not in operaciones:
            return {
                  "mensaje": f"error, tipo de operacion no soportada {operacion}",
                  "stock_actual": f"{inventario[producto]}"
            }
        if not isinstance(cantidad, int):
            return {
                  "mensaje": f"error, tipo de dato no adecuado o cantidad negativa"
            }  
        if cantidad <=0:
            return {
                  "mensaje": f"error, tipo de dato no adecuado o cantidad negativa"
            }  
        if operacion== "entrada":
             if producto not in inventario:
                inventario[producto]=cantidad
             else:
                  inventario[producto]+=cantidad
        if operacion== "salida" and producto in inventario and inventario[producto]>= cantidad:
             inventario[producto]=inventario[producto]-cantidad
                  

        print("comprobacion exitosa")

actualizar_inventario("inventario", "Mause",4 ,"Entrada")
```

```python
#6
#texto:{aaaaaabbbbbcc}
def comprimir(texto: str) -> str:
    resultado = []
    caracter_actual=texto[0]
    contador=1

    for caracter in texto[1:]:
        if caracter == caracter_actual:
            contador += 1
        else:
            resultado.append(f"{caracter_actual}{contador}")
            caracter_actual=caracter
            contador = 1
    resultado.append(f"{caracter_actual}{contador}")
    resultado="".join(resultado)
    return resultado
texto = 'aaabbcccdaa'
comprimir(texto)
```

```python
def descomprimir(texto:str)-> str:
    letras=texto[0::2]
    numeros=texto[1::2]
    resultado=[]
    for letra, numero in zip(letras,numeros):
        resultado.extend(
            letra*int(numero)
        )
    resultado="".join(resultado)
    return resultado
texto_encriptado= 'a3b2c3d1a2'
descomprimir(texto_encriptado)
```
