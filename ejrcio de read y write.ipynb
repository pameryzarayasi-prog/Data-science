```python
from typing import Any

def extract_text(ruta: str)-> dict[str, Any]:
    mi_lista=[]
    with open(ruta, "r") as f:
        contenido = f.read()
        lineas = contenido.split("\n")
        for linea in lineas:
            if "lista" in linea.lower():
                continue
            if not linea:
                continue

            if "instrucciones" in linea.lower():
                break
            mi_lista.append(linea)
    

    return mi_lista

        #print(linea)
        #print("-"*50)

extract_text(
    #"C:/Users/Usuario/Documents/data-science/data/estudiantes_python.txt"
    #"../data/estudiantes_python.txt"
    "C:/Users/Usuario/Documents/data-science/01-pyton-intro/data/estudiantes_python.txt"
)
```

```python
class Estudiante:
    def __init__(self, nombre:str, edad: int, ciudad: str, nota:float)->None:
        self.nombre=nombre
        self.edad=edad
        self.ciudad=ciudad
        self.nota=nota
class Aula:
    def __init__(self, estudiantes:list[str])->None:
        self.estudiante:list[Estudiante] =[]

        for estudiante in estudiantes:
            info=[item.strip() for item in estudiante.split(",")]
            estudiante= Estudiante(
                 nombre=info[0],
                 edad=int(info[1]),
                 ciudad=info[2],
                 nota=float(info[3])
            )

            self.estudiante.append(self.estudiante)

    def __len__(self)-> int:
        return len(self.estudiante)

    def filtro_ciudad(self, ciudad:str)->None:
        for estudiante in self.estudiante:
            if estudiante.ciudad.lower()==ciudad:
                print(estudiante.nombre)

    def get_aprobados(self,mi_nota: int=13)-> list[str]:
            estudiantes=[]
            for estudianate in self.estudiantes:
                if estudianate.nota >= mi_nota:
                    estudiantes.append(estudianate.nombre)
    def get_num_aproba(self,mi_nota: int=13)-> int:
        return len(self.get_aprobados(mi_nota)) 

    def export_alumnos(self, ruta:str)->None:
        with open(ruta,"w") as f:
            f.write("nombre, edad, ciudad, nota/n")

        for estudiante in self.estudiante:
            with open(ruta, "a") as f:
                f.write(f"{estudiante.nombre},{estudiante.edad},{estudiante.ciudad},{estudiante.nota}")


aula= Aula(contenido)



        
```
