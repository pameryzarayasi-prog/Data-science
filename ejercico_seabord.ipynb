```python
penguins_clean=penguins.dropna().copy()
penguins_clean.shape
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=penguins_clean,x="body_mass_g",bins=15)
plt.title("distribuscion de la masa")
plt.xlabel("masa corporal")
plt.ylabel("frecuencia")
```

```python
fig, axes = plt.subplots(1,2, figsize=(12,4))
sns.histplot(data=penguins_clean,x="body_mass_g",bins=6,ax=axes[0])
axes[0].set_title("25 bins")
axes[0].set_xlabel("masa corporal")

sns.histplot(data=penguins_clean,x="body_mass_g",bins=25,ax=axes[1])
axes[1].set_title("25 bins")
axes[1].set_xlabel("masa corporal")


plt.tight_layout
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=penguins_clean,x="flipper_length_mm",bins=18,kde=True)
plt.title("distribucion de la longitud")
plt.xlabel("longitud de la aleta")
plt.ylabel("frecuencia")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=penguins_clean,x="bill_length_mm",bins=18,hue="species", element="step")
plt.title("distribucion de la longitud pico")
plt.xlabel("longitud de pico")
plt.ylabel("frecuencia")
```

```python
plt.figure(figsize=[9,5])
sns.displot(data=penguins_clean,x="bill_depth_mm",bins=17,height=3,aspect=1.5)
plt.title("produndidad del pico")
plt.xlabel("Profundidad de pico")
```
