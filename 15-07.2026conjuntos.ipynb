```python
conjuntoA.union(conjuntoC)
```

```python
conjuntoA.intersection(conjuntoB)
```

```python
conjuntoA.difference(conjuntoB)
```

```python
conjuntoA.symmetric_difference(conjuntoB)
```

```python
conjuntoA | conjuntoB
```

```python
conjuntoA & conjuntoB
```

```python
conjuntoA - conjuntoB
```

```python
conjuntoA ^ conjuntoB
```
