```python
# 9. Histograma: longitud del sépalo
# Crea un histograma de `sepal_length_cm` con 10 bins, borde visible y etiquetas. Describe dónde se
# concentra la mayor parte de las observaciones.


plt.hist(df["sepal_length_cm"],bins=10, edgecolor="black" )
plt.title("longitud del sépalo")
plt.xlabel("longitud del sépalo")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# 10. Histograma y bins
# Crea dos histogramas de `petal_length_cm`: uno con 5 bins y otro con 20 bins. Colócalos lado a lado
# usando subplots. Explica brevemente cómo cambia la interpretación al modificar los bins.

fig, axes =plt.subplots(1,2, figsize=(12,9))

axes[0].hist(df["petal_length_cm"],bins=5, edgecolor="black", color="green" )
axes[0].set_title("longitud del petalo")
axes[0].set_ylabel("Frecuencia")
axes[0].set_xlabel("longitud del petalo cm")


axes[1].hist(df["petal_length_cm"],bins=20, edgecolor="black", color="pink" )
axes[1].set_title("longitud del petalo")
axes[1].set_ylabel("Frecuencia")
axes[1].set_xlabel("longitud del petalo cm")
```

```python
# 11. Boxplot por especie
# Construye un boxplot de `sepal_length_cm` separado por especie. Etiqueta cada caja con el nombre de
# su especie y compara visualmente las medianas.
fig, axes =plt.subplots(1,3, figsize=(12,4))

for ax, (species, gruop) in zip(axes,df.groupby("species")):
    sepal_length=gruop["sepal_length_cm"]
    ax.boxplot(sepal_length, patch_artist=True)
    ax.set_title(species)
    ax.set_xticks([1])
    ax.set_ylabel("longitud del sepalo")
fig.suptitle("longitud del sepalo por especie")
```

```python
# 12. Boxplot de variables numéricas
# Crea un solo boxplot que compare las cuatro variables numéricas del dataset. Usa etiquetas cortas y
# legibles para cada caja. ¿Qué problema de escala observas?
plt.figure(figsize=(10,5))
grupos = [df[columna] for columna in df.select_dtypes("number").columns]

Columnas_numericas = df.select_dtypes("number").columns
grupos

plt.boxplot(grupos,tick_labels=Columnas_numericas)
plt.xlabel("Especies")
plt.ylabel("Medida")
plt.title('variables numéricas del dataset Iris')
plt.show()
```

```python
# 13. Subplots 1x2
# Crea una figura con dos subplots: a la izquierda un scatter de `sepal_length_cm` vs. `sepal_width_cm`; a
# la derecha un histograma de `sepal_length_cm`. Agrega un título independiente a cada panel.

fig, axes =plt.subplots(1,2, figsize=(12,9))

axes[0].scatter(df["sepal_length_cm"],df["sepal_width_cm"], edgecolor="black", color="grey" )
axes[0].set_title("Longitud del sepalo vs. peso del sepalo")
axes[0].set_ylabel("Peso del sepalo")
axes[0].set_xlabel("Longitud del sepalo cm")


axes[1].hist(df["sepal_length_cm"],bins=20, edgecolor="black", color="pink" )
axes[1].set_title("longitud del sepalo")
axes[1].set_ylabel("Frecuencia")
axes[1].set_xlabel("longitud del sepalo cm")
```
