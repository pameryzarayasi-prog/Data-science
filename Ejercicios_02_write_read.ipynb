```python
from typing import Any

def extrae(ruta: str)-> dict[str, Any]:
    mi_lista=[]
    with open(ruta, "r") as f:
        informacion = f.read()
        lineas = informacion.split("/n")
        for linea in lineas:
            if "registro" in linea.lower():
                continue
            if not linea:
                continue
            
            if "instrucciones" in linea.lower():
                break
            mi_lista.append(linea)

extrae(
   
    "C:/Users/Usuario/Documents/data-science/01-pyton-intro/data/ventas_polos_aqp.txt"
)
```
