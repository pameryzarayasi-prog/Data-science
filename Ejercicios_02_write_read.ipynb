```python
#1. Lee el archivo línea por línea.
from typing import Any

def extrae(ruta: str)-> dict[str, Any]:
    mi_lista=[]
    with open(ruta, "r") as f:
        informacion = f.read()
        lineas = informacion.split("\n")

        for linea in lineas:
            if "registro" in linea.lower() or "id, producto" in linea.lower():
                continue
            if not linea.strip():
                continue
            
            if "instrucciones sugeridas" in linea.lower():
                break

            mi_lista.append(linea)

    return mi_lista

informacion=extrae(
   
    "C:/Users/Usuario/Documents/data-science/01-pyton-intro/data/ventas_polos_aqp.txt"
)
```

```python
#4. Calcula el total vendido por cada registro: cantidad * precio_unitario.
class Venta:
    def __init__(self, ID:str, producto: str, talla: str, color:str, cliente: str, ciudad: str, cantidad:int, precio_unitario:float, metodo_pago:str, estado:str )->None:
        self.id=ID
        self.producto=producto
        self.talla=talla
        self.color=color
        self.cliente=cliente
        self.ciudad=ciudad
        self.cantidad=cantidad
        self.precio_unitario=precio_unitario
        self.metodo_pago=metodo_pago
        self.estado=estado
        self.total_vendido=self.cantidad*self.precio_unitario
        

class Ventas:

    def __init__(self, informacion: list[str]) -> None:
        self.ventas = []

        for linea in informacion:
            datos = linea.split(",")

            venta = Venta(
                datos[0].strip(),
                datos[1].strip(),
                datos[2].strip(),
                datos[3].strip(),
                datos[4].strip(),
                datos[5].strip(),
                int(datos[6].strip()),
                float(datos[7].strip()),
                datos[8].strip(),
                datos[9].strip()
            )

            self.ventas.append(venta)

    def total_vendido(self):
        for venta in self.ventas:
            total = round(venta.cantidad * venta.precio_unitario,1)
            print(f"El total vedido es {venta.id} = {total} soles")

    def estado_producto(self, estado: str):
        estado_producto = []

        for venta in self.ventas:
            if venta.estado.strip().lower() == estado.lower().strip():
                print(venta.id, venta.cliente, venta.producto)
            estado_producto.append(venta)

        return estado_producto


ventas = Ventas(informacion)
ventas.total_vendido()

ventas.estado_producto("Entregado")
```
