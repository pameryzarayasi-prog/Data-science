```sql
CREATE TABLE IF NOT EXISTS tienda.categorias (
    id      INT     NOT NULL,
    nombre  STRING  NOT NULL
);

CREATE TABLE IF NOT EXISTS tienda.productos (
    id      INT     NOT NULL,
    nombre  STRING  NOT NULL,
    precio  NUMERIC NOT NULL,
    stock   INT,
    categoria_id INT  NOT NULL
);

CREATE TABLE IF NOT EXISTS detalle_pedidos (
    pedido_id INT NOT NULL,
    producto_id INT NOT NULL,
    cantidad INT,
    precio_unit NUMERIC
);

CREATE TABLE IF NOT EXISTS pedidos (
    id INT NOT NULL,
    cliente_id INT NOT NULL,
    fecha DATE,
    total NUMERIC
);

CREATE TABLE IF NOT EXISTS clientes (
    id INT NOT NULL,
    nombre STRING NOT NULL,
    email STRING,
    pais STRING,
    telefono STRING
);
```

```sql
select 
    *,
    length(nombre) as longitud_nombre,
    upper(nombre) as nombre_mayusculas,
    lower(nombre) as nombre_minusculas,
    concat(nombre, ' ;', email) as nombre_email,
    ' UNSA  ' AS universidad,
    length(' UNSA  ') AS longitud_universidad,
    TRIM(' UNSA  ') AS universidad_trim,
    LENGTH(' UNSA  ') AS longitud_universidad_trim,
    replace(pais,'ES','España') as
    pais_sin_espacios,
    replace(nombre,'o','x') as
    nombre_reemplazo_o_x,
    substring(nombre,1,3) as nombre_tres_primeros_caracteres
from clientes;
```

```sql
select 
    'oreo' as galleta,
    4.5 as precio,
    round (4.595,1) as precio_redondeado,
    ROUND(4.5) AS precio_original,
    ceil(4.5) AS precio_original_techo,
    floor(4.5) AS precio_original_piso,
    abs(-4.5) as precio_positivo,
    sqrt(16) as raiz_cuadrada,
    pow(2,3) as potencia
```

```sql
select  
    pais,
    case
        when pais = 'ES' then 'España'
        when pais = 'FR' then 'Francia'
        when pais = 'IT' then 'Italia'
        when pais = 'MX' then 'Mexico'
        when pais = 'PE' then 'Perú'
        when pais = 'CO' then 'Colombia'
        when pais = 'CL' then 'Chile'
        else 'Argentina'
    end as pais_descripcion
from clientes
```

```sql
select 
    nombre,
    precio,
    case
        when precio > 300 then 'caro'
        when precio > 50 then 'medio'
        else 'barato'
    end as clasificacion
from productos
```

```sql
select *,
    coalesce(catergorias_id, 0) as categoria_id_imputada
from productos;
```
