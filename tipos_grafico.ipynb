```python
from matplotlib import pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,20,25,30,45]


plt.plot(x,y, marker="o", linestyle="-", color="b")
plt.xlabel("Perido")
plt.ylabel("Valor")
plt.title("Ejemplo de grafico de lineas")
plt.show()
```

```python
categorias=["A","B","C","D"]
valores= [40,70,30,80]

plt.bar(categorias,valores, color=["red","green","black","blue"])
plt.xlabel("Categorias")
plt.ylabel("Frecuencia")
plt.title("Grafico de barras verticales")
plt.show()
```

```python
plt.barh(categorias,valores, color="#B628FA")
plt.xlabel("Categorias")
plt.ylabel("Frecuencia")
plt.title("Grafico de barras verticales")
plt.show()
```

```python
frutas=["Manzana","Platano","Peras"]
porcentaje=[30,50,20]

plt.pie(porcentaje, labels=frutas, autopct='%1.1f%%')
plt.title("Grafico de pastel")
plt.show()
```

```python
import numpy as np

x= np.random.rand(50)
y= np.random.rand(50)

plt.scatter(x,y, color='blue')
plt.xlabel("Variable (x)")
plt.ylabel("Variable (y)")
plt.title("Grafico de dispersion")
plt.show()
```

```python
datos= np.random.rand(1000)
plt.hist(datos, bins=20, color="orange")
plt.xlabel("Variable (x)")
plt.ylabel("Variable (y)")
plt.title("Grafico de dispersion")
plt.show()
```
