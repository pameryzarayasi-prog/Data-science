```python
# Ejercicio 2. Inventario después de las ventas
stock = np.array([
    [25, 18, 30, 12, 20],
    [14, 22, 16, 28, 19],
    [35, 17, 24, 15, 21]])

ventas = np.array([ 
    [10, 20, 12, 8, 7],
    [16, 10, 18, 9, 20],
    [15, 19, 8, 17, 11]])

#1
stock_restante=stock-ventas
print(stock_restante)
#2
mascara=stock_restante<0
print(stock_restante[mascara])


#3
stock_restante_corregido=stock_restante.copy()
stock_restante_corregido[mascara]=0
print(stock_restante_corregido)

#4
stock_restante_antes=stock_restante.sum(axis=1)
print(stock_restante_antes)
```

```python
# Ejercicio 3. Facturación por sucursal
precios = np.array([
    [4.5, 6.0, 3.5, 8.0],
    [4.7, 5.8, 3.6, 8.2],
    [4.6, 6.1, 3.4, 7.9]])
unidades = np.array([
    [30, 18, 42, 15],
    [25, 20, 35, 17],
    [28, 22, 40, 14]])

total=precios*unidades
print(total)
print("-"*20)

#2
total_sucursal=total.sum(axis=1)
total_productos=total.sum(axis=0)
print(total_sucursal)
print(total_productos)
print("-"*20)

#3
print(total_sucursal.argmax())

#4
valor_max=total.max()
print(valor_max)

mascara=total
for fila in range(mascara.shape[0]):
    for columna in range(mascara.shape[1]):
        if valor_max==total[fila,columna]:
            print(fila,columna)

print(total.T)
```
