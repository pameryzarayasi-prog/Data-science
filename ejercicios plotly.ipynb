```python
# Construye un gráfico de dispersión que muestre la relación entre `sepal_length` y `sepal_width`. Diferencia
# cada observación según la especie mediante el parámetro `color`. Agrega un título descriptivo y etiquetas
# claras para ambos ejes.
# Conceptos a practicar: px.scatter, ejes, color y títulos.

fig=px.scatter(
    df,
    x="sepal_length",
    y="sepal_width",
    color="species",
    title="longitud vs ancho del sepalo",
    labels={
        "sepal_length": "longitud del sepalo",
        "sepal_width": "Ancho del sepalo",
        "species": "especies"

    }
)
fig.show()
```

```python
#2
fig=px.histogram(
    data_frame=df,
    x="sepal_length",
    color="species",
    nbins=20,
    barmode="overlay",
    title="longitud vs ancho del sepalo",
    labels={
        "sepal_length": "longitud del sepalo",
        "species": "especies"

    }
)
fig.show()
```

```python
# 3. Boxplot de ancho del pétalo
# Crea un boxplot de `petal_width` para cada especie. Muestra también los puntos individuales de las
# observaciones. Identifica visualmente qué especie presenta los valores más altos y cuál tiene menor
# dispersión.
# Conceptos a practicar: px.box, distribución, mediana, IQR y puntos
fig=px.box(
    data_frame=df,
    x="species",
    y="petal_width",
    color="species",
    points="all",
    title="ancho de petalo por especie",
    labels={"specie": "especie","petal_width": "ancho del petalo"}
)
fig.show()
```

```python
# 4. Violin plot del largo del sépalo
# Construye un gráfico de violín para comparar `sepal_length` entre especies. Incluye dentro del violín un
# boxplot y muestra todos los puntos. Explica qué información adicional aporta el violín respecto al boxplot.
# Conceptos a practicar: px.violin, densidad, box y puntos
fig=px.violin(
    data_frame=df,
    x="species",
    y="petal_width",
    color="species",
    points="all",
    title="ancho de petalo por especie",
    labels={"specie": "especie","petal_width": "ancho del petalo"}
)
fig.show()
```

```python
# 5. Promedio de largo del sépalo por especie
# Calcula el promedio de `sepal_length` por especie con Pandas y visualízalo mediante un gráfico de barras.
# Ordena las especies de menor a mayor promedio y muestra el valor promedio sobre cada barra.
# Conceptos a practicar: groupby + px.bar + text_auto.
promedio_sepalo=(
    df.groupby("species", as_index=False)["sepal_length"].mean()
)

fig=px.bar(
    promedio_sepalo,
    x="species",
    y="sepal_length",
    text_auto=".3f",
    title="promedio de longitud",
    labels={"species":"especie","sepal_length":"longitud de petalo"}
)
fig.show()
```

```python
# 6. Comparación de promedios de las cuatro medidas
# Calcula, para cada especie, el promedio de `sepal_length`, `sepal_width`, `petal_length` y `petal_width`.
# Transforma el resultado a formato largo y crea un gráfico de barras agrupadas que permita comparar las
# cuatro variables entre especies.
# Conceptos a practicar: groupby, melt y barras agrupadas.

promedios= df.groupby("species", as_index=False)[["sepal_length", "sepal_width",'petal_length', 'petal_width']].mean()
promedios
fig=px.bar(
    promedios,
    x="species",
    y=["sepal_length", "sepal_width",'petal_length', 'petal_width'],
    barmode="group",
    title="promedio de longitud",
    labels={"species":"especie","sepal_length":"longitud de petalo"}
)
fig.show()

```
