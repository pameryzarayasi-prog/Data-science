```sql
--EJERCIO 1
select 
nombre,
precio,
    case
        when precio > 300 then 'caro'
        when precio > 50 then 'medio'
        else 'barato'
    end as clasificacion
from productos;
```

```sql
--EJERCICIO 2
SELECT *
FROM clientes
where pais IN('ES','MX') AND nombre like 'M%'
```

```sql
-- EJERCICO 3
SELECT*
FROM productos
where precio BETWEEN 20 AND 60;
```

```sql
--EJERCIO 4
SELECT *, 
    coalesce(telefono,null,'sin dato') as telefono_coalesce
FROM clientes;
--WHERE telefono IS NULL;
```

```sql
--EJERCIO 5
select *,
    precio*1.21 as precio_con_iva
from productos

order by precio_con_iva desc
limit 10
    
```
